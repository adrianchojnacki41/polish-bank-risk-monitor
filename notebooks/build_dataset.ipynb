```python
def read_dataset(REFRESH):
    
    fx = clean_fx(load_fx(REFRESH))
    validate_fx(fx)
    
    stocks = clean_stocks(load_stocks(REFRESH))
    validate_stocks(stocks)
    return fx,stocks
```
